```python
def load_json():
       with open("conversation.json","r" ,encoding="utf-8") as f:
          conversation= json.load(f)
       return conversation


def plot_line_chart(sentiments):
    x = list(range(1, len(sentiments) + 1))  
    y = sentiments                           

    plt.plot(x, y, marker='o')
    plt.xlabel("The number of message in the conversation")
    plt.ylabel("Sentiment score")
    plt.title("Change in sentiment throughout the conversation")
    plt.grid(True)
    plt.show()

def lineChart():
    for_lineChart = []
    conversations = load_json()

    conversation = conversations[-1]
   
   
    for message in conversation["messages"]:
        sentiment = calculate_sentiment(message["user"])
        for_lineChart.append(sentiment)

    plot_line_chart(for_lineChart)

        

   
    
    
```

```python
import matplotlib.pyplot as plt

def tone_compare():

    gen = conversations_generator()
    conversation = next(gen)

    bot_add = 0
    user_add = 0
    bot_count = 0
    user_count = 0

   
    for message in conversation["messages"]:
        sentiment = calculate_sentiment(message["user"])
        if sentiment is not None:
         user_add += sentiment
         user_count += 1

   
    for message in conversation["messages"]:
        sentiment = calculate_sentiment(message["model"])
        if sentiment is not None:
         bot_add += sentiment
         bot_count += 1

    bot_avg = bot_add / bot_count if bot_count > 0 else 0
    user_avg = user_add / user_count if user_count > 0 else 0

    
    labels = ['User', 'Bot']
    averages = [user_avg, bot_avg]

    plt.figure(figsize=(6, 4))
    plt.bar(labels, averages, color=['skyblue', 'orange'])
    plt.title("Comparing tone between user and bot")
    plt.ylim(-1, 1)  
    plt.ylabel("Sentiment Average")
    plt.show()
```

```python
tone_compare()
```
